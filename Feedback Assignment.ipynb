## Feedback

```python
def write_feedback():
    fb = input("Enter The Feedback: ")
    good = ["good", "great", "excellent", "awesome", "amazing", "love", "best", "perfect","nice", "happy", "fantastic", "outstanding", "wonderful", "superb", "brilliant",
    "impressive", "helpful", "fast", "smooth", "satisfied", "pleased", "top-notch","recommend", "delighted"]
    bad = ["bad", "worst", "terrible", "horrible", "poor", "hate", "slow", "useless","disappointed", "issue", "broken", "annoying", "awful", "defective", "error",
    "failed", "frustrating", "buggy", "unacceptable", "unhelpful", "problem","glitch", "refund", "rubbish"]
    try:        
        if any(word in fb.lower() for word in good):
            with open("Good_Feedback.txt","a") as gf:
                gf.write(fb+"\n")
        elif any(word in fb.lower() for word in bad):
            with open("Bad_Feedback.txt","a") as bf:
                bf.write(fb+"\n")
        else:
            with open("Neutral_Feedback.txt","a") as nf:
                nf.write(fb+"\n")
    except Exception as e:
        print("An Error Occured while writing the file")
    else:
        print("Thank you for your feedback!")
    finally:
        print("Feedback submission process completed.")

def read_feedback():
    try:
        with open("Good_Feedback.txt","r") as gf:
            Good_Feedback = gf.readlines()
        with open("Bad_Feedback.txt","r") as bf:
            Bad_Feedback = bf.readlines()
        with open("Neutral_Feedback.txt","r") as nf:
            Neutral_Feedback = nf.readlines()

        print("Good_Feedback:")
        for feedback in Good_Feedback:
            print(feedback.strip())
        print("Bad_Feedback:")
        for feedback in Bad_Feedback:
            print(feedback.strip())
        print("Neutral_Feedback:")
        for feedback in Neutral_Feedback:
             print(feedback.strip())
    except Exception as e:
        print("An Error Occured while reading the file")
    else:
        print("Feedback reading process completed")
    finally:
        print("Feedback submission process completed")
```

```python
write_feedback()
```
